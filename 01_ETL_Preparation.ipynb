Este cuaderno limpia y prepara el dataset de créditos rurales para el modelado.

```python
import pandas as pd
import numpy as np
import os
```

## 1. Carga de Datos

```python
df = pd.read_csv('data/creditos_rurales.csv')
print(f"Dataset cargado: {df.shape[0]} filas, {df.shape[1]} columnas")
print(f"\nNulos totales: {df.isnull().sum().sum()}")
df.head()
```

## 2. Análisis de Nulos

```python
nulos_antes = df.isnull().sum()
print("Nulos por columna (ANTES):")
print(nulos_antes[nulos_antes > 0])
```

## 3. Imputación de Valores Nulos

```python
# Cédula: rellenar con 'SIN_CEDULA'
if 'Cedula' in df.columns:
    df['Cedula'] = df['Cedula'].fillna('SIN_CEDULA')

# Experiencia: rellenar con mediana (robusto ante atípicos)
if 'Experiencia_Anios' in df.columns:
    mediana_exp = df['Experiencia_Anios'].median()
    df['Experiencia_Anios'] = df['Experiencia_Anios'].fillna(mediana_exp)
    print(f"Experiencia: mediana = {mediana_exp}")

# Garantía: rellenar con moda (categórica binaria)
if 'Garantia_Respaldada' in df.columns:
    moda_gar = df['Garantia_Respaldada'].mode()[0]
    df['Garantia_Respaldada'] = df['Garantia_Respaldada'].fillna(moda_gar)
    print(f"Garantía: moda = {moda_gar}")
```

## 4. Conversión de Tipos

```python
# Convertir a enteros
df['Experiencia_Anios'] = df['Experiencia_Anios'].astype(int)
df['Garantia_Respaldada'] = df['Garantia_Respaldada'].astype(int)
df['Tiene_Energia_Solar'] = df['Tiene_Energia_Solar'].astype(int)
df['Tiene_Riego'] = df['Tiene_Riego'].astype(int)
df['Exporta_Productos'] = df['Exporta_Productos'].astype(int)

print("Tipos de datos convertidos correctamente")
```

## 5. Validación de Límites

```python
# Limitar valores a rangos lógicos
df['Tasa_Interes_EA'] = df['Tasa_Interes_EA'].clip(6.0, 31.0)
df['Monto_Prestamo_COP'] = df['Monto_Prestamo_COP'].clip(5000000, None)
df['Edad'] = df['Edad'].clip(18, 75)

print("Límites aplicados:")
print(f"  Tasa: {df['Tasa_Interes_EA'].min():.2f}% - {df['Tasa_Interes_EA'].max():.2f}%")
print(f"  Monto: ${df['Monto_Prestamo_COP'].min():,.0f} - ${df['Monto_Prestamo_COP'].max():,.0f}")
print(f"  Edad: {df['Edad'].min()} - {df['Edad'].max()}")
```

## 6. Verificación Final

```python
nulos_despues = df.isnull().sum().sum()
print(f"Nulos eliminados: {df.isnull().sum().sum()} -> {nulos_despues}")
print(f"\nDataset final: {df.shape[0]} filas, {df.shape[1]} columnas")
print(f"\nColumnas: {list(df.columns)}")
```

## 7. Guardar Dataset Limpio

```python
df.to_csv('data/creditos_rurales_limpio.csv', index=False)
print(f"Dataset limpio guardado: data/creditos_rurales_limpio.csv")
```
