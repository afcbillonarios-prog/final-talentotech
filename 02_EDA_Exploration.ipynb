Este cuaderno analiza las distribuciones, correlaciones y patrones del dataset de créditos rurales.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('viridis')
%matplotlib inline
```

## 1. Carga de Datos Limpios

```python
df = pd.read_csv('data/creditos_rurales_limpio.csv')
print(f"Dataset: {df.shape[0]} filas, {df.shape[1]} columnas")
df.head()
```

## 2. Estadísticas Descriptivas

```python
df.describe()
```

## 3. Distribución por Sector Productivo

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distribución de sectores
df['Sector_Productivo'].value_counts().plot(kind='bar', ax=axes[0], color='viridis')
axes[0].set_title('Distribución por Sector Productivo', fontsize=14)
axes[0].set_ylabel('Cantidad')
axes[0].tick_params(axis='x', rotation=45)

# Tasa promedio por sector
df.groupby('Sector_Productivo')['Tasa_Interes_EA'].mean().sort_values().plot(kind='barh', ax=axes[1], color='viridis')
axes[1].set_title('Tasa Promedio por Sector', fontsize=14)
axes[1].set_xlabel('Tasa E.A. (%)')

plt.tight_layout()
plt.show()
```

## 4. Distribución de la Variable Objetivo

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histograma
df['Tasa_Interes_EA'].hist(bins=30, ax=axes[0], color='green', edgecolor='black')
axes[0].set_title('Distribución de Tasa de Interés', fontsize=14)
axes[0].set_xlabel('Tasa E.A. (%)')
axes[0].set_ylabel('Frecuencia')

# Boxplot
df['Tasa_Interes_EA'].plot(kind='box', ax=axes[1])
axes[1].set_title('Boxplot de Tasa de Interés', fontsize=14)

plt.tight_layout()
plt.show()
```

## 5. Correlación de Variables

```python
numeric_cols = ['Edad', 'Experiencia_Anios', 'Hectareas', 'Ingresos_Mensuales_COP',
                'Monto_Prestamo_COP', 'Plazo_Meses', 'Garantia_Respaldada',
                'Subsidio_Gobierno', 'Tecnologias_Usadas', 'Tiene_Energia_Solar',
                'Tiene_Riego', 'Tasa_Interes_EA']

corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Matriz de Correlación', fontsize=16)
plt.tight_layout()
plt.show()
```

## 6. Impacto de Variables en la Tasa

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Garantía vs Tasa
df.groupby('Garantia_Respaldada')['Tasa_Interes_EA'].mean().plot(kind='bar', ax=axes[0,0], color=['red', 'green'])
axes[0,0].set_title('Impacto de Garantía en Tasa')
axes[0,0].set_ylabel('Tasa E.A. (%)')

# Subsidio vs Tasa
df.groupby('Subsidio_Gobierno')['Tasa_Interes_EA'].mean().plot(kind='bar', ax=axes[0,1], color=['red', 'green'])
axes[0,1].set_title('Impacto de Subsidio en Tasa')
axes[0,1].set_ylabel('Tasa E.A. (%)')

# Energía Solar vs Tasa
df.groupby('Tiene_Energia_Solar')['Tasa_Interes_EA'].mean().plot(kind='bar', ax=axes[1,0], color=['red', 'green'])
axes[1,0].set_title('Impacto de Energía Solar en Tasa')
axes[1,0].set_ylabel('Tasa E.A. (%)')

# Tecnologías vs Tasa
df.groupby('Tecnologias_Usadas')['Tasa_Interes_EA'].mean().plot(kind='bar', ax=axes[1,1], color='viridis')
axes[1,1].set_title('Impacto de Tecnologías en Tasa')
axes[1,1].set_ylabel('Tasa E.A. (%)')

plt.tight_layout()
plt.show()
```

## 7. Distribución por Departamento

```python
plt.figure(figsize=(12, 6))
df['Departamento'].value_counts().head(15).plot(kind='bar', color='viridis')
plt.title('Top 15 Departamentos con Más Créditos', fontsize=14)
plt.ylabel('Cantidad de Créditos')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 8. Producción por Sector

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Producción total
df.groupby('Sector_Productivo')['Produccion_Anual_Ton'].sum().sort_values(ascending=False).plot(kind='bar', ax=axes[0], color='viridis')
axes[0].set_title('Producción Total por Sector (Toneladas)', fontsize=14)
axes[0].set_ylabel('Toneladas/Año')

# Exportaciones
df.groupby('Sector_Productivo')['Exporta_Productos'].sum().sort_values(ascending=False).plot(kind='bar', ax=axes[1], color='viridis')
axes[1].set_title('Exportadores por Sector', fontsize=14)
axes[1].set_ylabel('Cantidad')

plt.tight_layout()
plt.show()
```

## Resumen del EDA

- **Distribución de sectores:** Agricultura (30%), Ganadería (20%), Piscicultura (15%)
- **Tasa promedio:** ~16.4% E.A.
- **Variables con mayor impacto:** Subsidio (-5.5%), Garantía (-3.2%), Energía Solar (-1.0%)
- **Correlaciones:** La tasa tiene correlación negativa con subsidio, garantía y tecnología
- **Producción:** Más de 15,000 toneladas/año de alimentos
